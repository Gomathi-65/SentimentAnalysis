```python
import re
import nltk
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
from nltk.corpus import wordnet



stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
negators = {"no","nor","not","don't","ain","aren","couldn","didn","doesn","hadn","hasn","haven","isn","mightn","mustn","needn","shan","shouldn","wasn","weren","won","wouldn"}
stop_words = stop_words - negators

def _to_wn(pos):
  return {"J": wordnet.ADJ, "V": wordnet.VERB, "N": wordnet.NOUN, "R": wordnet.ADV}.get(pos[0], wordnet.NOUN)

def preprocess_text(text):

    if pd.isnull(text):
      return ""
    # Convert to lowercase
    text = text.lower()
    # Remove special characters and numbers
    text = re.sub(r"[^a-zA-Z\s ']", " ", text)
    #collapse extra spaces
    text = re.sub(r"\s+", " ", text)
    # Tokenize
    tokens = text.split()
    #remove stopwords but keep negators
    tokens = [token for token in tokens if token not in stop_words or token in negators]
    # Lemmatize
    tagged_tokens = pos_tag(tokens)
    lemmatized_tokens = [lemmatizer.lemmatize(token, _to_wn(pos)) for token, pos in tagged_tokens]
    return " ".join(lemmatized_tokens)

df['cleaned_reviews']= df['review'].apply (preprocess_text)
```

```python
#TEXT PREPROCESSING SETUP
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

negators = {"no","nor","not","don't","didn","doesn","isn","wasn","weren"}
stop_words = stop_words - negators

def _to_wn(pos):
    return {
        "J": wordnet.ADJ,
        "V": wordnet.VERB,
        "N": wordnet.NOUN,
        "R": wordnet.ADV
    }.get(pos[0], wordnet.NOUN)

def preprocess_text(text):
    if pd.isnull(text):
        return ""
    text = text.lower()
    text = re.sub(r"[^a-zA-Z\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    tokens = text.split()
    tokens = [t for t in tokens if t not in stop_words or t in negators]
    tagged = pos_tag(tokens)
    lemmatized = [lemmatizer.lemmatize(w, _to_wn(p)) for w, p in tagged]
    return " ".join(lemmatized)

df["cleaned_reviews"] = df["review"].apply(preprocess_text)
```

```python
import pandas as pd
import numpy as np
import re
import joblib
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
from nltk.corpus import wordnet

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.utils import resample
from sklearn.model_selection import GridSearchCV

from sklearn.metrics import classification_report, accuracy_score
```

```python
#LOAD FOR MODEL TRAINING
df = pd.read_csv("processed_cleaned_reviews.csv")
```

```python
df = df[["cleaned_reviews", "sentiment"]].dropna()

print("\nBefore balancing:\n", df["sentiment"].value_counts())
```

```python
# BALANCE DATA USING UPSAMPLING
max_count = df["sentiment"].value_counts().max()
balanced_parts = []

for label, group in df.groupby("sentiment"):
    if len(group) < max_count:
        up = resample(group, replace=True, n_samples=max_count, random_state=42)
        balanced_parts.append(up)
    else:
        balanced_parts.append(group)

df_balanced = pd.concat(balanced_parts).sample(frac=1, random_state=42).reset_index(drop=True)

print("\nAfter balancing:\n", df_balanced["sentiment"].value_counts())
     
```

```python
# HYBRID SENTIMENT FIX (MODEL + RULE BASED)
# SIMPLE SENTIMENT LEXICON
positive_words = {
    "good", "great", "excellent", "awesome", "nice", "amazing",
    "perfect", "love", "super", "best", "fantastic", "wonderful",
    "better", "fast", "smooth", "happy"
}

negative_words = {
    "bad", "worst", "poor", "slow", "terrible", "awful",
    "problem", "issue", "bug", "crash", "fail", "failed",
    "error", "lag", "laggy", "hate"
}

neutral_words = {
    "ok", "okay", "average", "fine", "normal", "decent", "regular"
}
```

```python
 #TRAIN TEST SPLIT

X = df_balanced["cleaned_reviews"]
y = df_balanced["sentiment"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
     
```

```python
#TF-IDF VECTORIZATION
vectorizer = TfidfVectorizer(max_features=20000, ngram_range=(1,2))
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
```

```python
#LOGISTIC REGRESSION MODEL
model = LogisticRegression(max_iter=1000, solver="lbfgs")
model.fit(X_train_vec, y_train)
```

```python
cv_scores = cross_val_score(model, X_train_vec, y_train, cv=5, scoring="accuracy")
print("\nCV Accuracy:", cv_scores.mean())
```

```python
#EVALUATION
y_pred = model.predict(X_test_vec)

print("\nCLASSIFICATION REPORT:\n", classification_report(y_test, y_pred))
print("CONFUSION MATRIX:\n", confusion_matrix(y_test, y_pred))
```

```python
#SAVE MODEL & VECTORIZER
joblib.dump(model, "text_classifier_balanced.joblib")
joblib.dump(vectorizer, "vectorizer_balanced.joblib")

print("\n✅ Saved Model & Vectorizer")
```

```python
#Hyperparameter tunning

param_grid = {
    'C': [0.01, 0.1, 1, 10, 100],          # Regularization strength
    'penalty': ['l2'],                    # Best for text data
    'solver': ['lbfgs', 'liblinear'],     # Reliable solvers
    'max_iter': [100, 300, 500]
}
grid = GridSearchCV(
    LogisticRegression(),
    param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=2
)

grid.fit(X_train_vec, y_train)


best_model = grid.best_estimator_
best_params = grid.best_params_

print("Best Parameters:", best_params)
y_pred = best_model.predict(X_test_vec)

print("✅ Accuracy:", accuracy_score(y_test, y_pred))

print("\n✅ CLASSIFICATION REPORT:\n")
print(classification_report(y_test, y_pred))

print("\n✅ CONFUSION MATRIX:\n")
cm = confusion_matrix(y_test, y_pred)
print(cm)

```

```python
# SMART HYBRID SENTIMENT PREDICTION (FIXED)
positive_words = {
    "good", "great", "excellent", "awesome", "nice", "amazing", "perfect",
    "love", "super", "best", "fantastic", "wonderful", "better", "fast",
    "smooth", "happy"
}

negative_words = {
    "bad", "worst", "poor", "slow", "terrible", "awful",
    "problem", "issue", "bug", "crash", "crashes",
    "fail", "failed", "error", "lag", "laggy", "hate","not good"
}

neutral_words = {
    "ok", "okay", "average", "fine", "normal", "decent", "regular"
}

def smart_predict(text):
    cleaned = preprocess_text(text)
    words = cleaned.split()

    pos_hits = sum(1 for w in words if w in positive_words)
    neg_hits = sum(1 for w in words if w in negative_words)
    neu_hits = sum(1 for w in words if w in neutral_words)

    # ✅ Rule-based first (MOST IMPORTANT FIX)
    if pos_hits > neg_hits:
        return "positive ✅ (rule-based)"
    elif neg_hits > pos_hits:
        return "negative ❌ (rule-based)"
    elif neu_hits > 0:
        return "neutral ⚪ (rule-based)"

    # ✅ Very short text default
    if len(words) <= 2:
        return "neutral ⚪ (short-text default)"

    # ✅ ML Model fallback
    vec = vectorizer.transform([cleaned])
    pred = model.predict(vec)[0]
    return pred + " 🤖 (model)"
```

```python
#LIVE SENTENCE TESTING

while True:
    sent = input("\nEnter sentence (or type exit): ")
    if sent.lower().strip() == "exit":
        break
    print("Sentiment =", smart_predict(sent))
```
